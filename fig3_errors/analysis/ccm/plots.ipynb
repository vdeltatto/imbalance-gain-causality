```python
import numpy as np
import matplotlib.pyplot as plt
import pickle
import scienceplots

#plt.style.use('science')
colors = ["#00BBF9", "#CC2836", "#F3A712","#3eb690","#A05496","#114b5f"]
markers = ['o-','s-','^-']
```

### Rossler, identical

```python
epsilons = np.linspace(0.,0.25,30)
Ls = np.linspace(100,30000,50,dtype=int)

ccm_X_to_Y = np.zeros((20,len(epsilons),len(Ls)))
ccm_Y_to_X = np.zeros((20,len(epsilons),len(Ls)))

for seed in range(20):
    ccm_X_to_Y[seed], ccm_Y_to_X[seed] = pickle.load(open(f"./pickles/rossler_id/seed{seed}.p","rb"))
print(ccm_X_to_Y.shape, ccm_Y_to_X.shape)

ccm_X_to_Y_avg = ccm_X_to_Y.mean(axis=0)[:,-1]
ccm_X_to_Y_err = ccm_X_to_Y.std(axis=0)[:,-1] / ccm_X_to_Y.shape[0]

ccm_Y_to_X_avg = ccm_Y_to_X.mean(axis=0)[:,-1]
ccm_Y_to_X_err = ccm_Y_to_X.std(axis=0)[:,-1] / ccm_Y_to_X.shape[0]
```

```python
plt.plot(epsilons, ccm_X_to_Y_avg, markers[0], color=colors[0], label="$X\\rightarrow Y$")
plt.plot(epsilons, ccm_Y_to_X_avg, markers[1], color=colors[1], label="$Y\\rightarrow X$")
plt.fill_between(epsilons, ccm_X_to_Y_avg - ccm_X_to_Y_err,
                           ccm_X_to_Y_avg + ccm_X_to_Y_err, color=colors[0], alpha=0.4)
plt.fill_between(epsilons, ccm_Y_to_X_avg - ccm_Y_to_X_err,
                           ccm_Y_to_X_avg + ccm_Y_to_X_err, color=colors[1], alpha=0.4)
plt.ylabel("Imbalance Gain [%]")
#plt.set_xticks([])
#plt.legend()
```
