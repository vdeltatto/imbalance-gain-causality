```python
import numpy as np
import matplotlib.pyplot as plt
import pickle
import scienceplots

#plt.style.use('science')
colors = ["#00BBF9", "#CC2836", "#F3A712","#3eb690","#A05496","#114b5f"]
markers = ['o-','s-','^-']
```

### Rossler, identical

```python
epsilons = np.linspace(0.,0.25,30)

egc_index_X_to_Y = np.zeros((20,len(epsilons)))
egc_index_Y_to_X = np.zeros((20,len(epsilons)))

for seed in range(20):
    egc_index_X_to_Y[seed], egc_index_Y_to_X[seed] = pickle.load(open(f"./pickles/rossler_id/seed{seed}.p","rb"))
print(egc_index_X_to_Y.shape, egc_index_Y_to_X.shape)

egc_index_X_to_Y_avg = egc_index_X_to_Y.mean(axis=0)
egc_index_X_to_Y_err = egc_index_X_to_Y.std(axis=0) / egc_index_X_to_Y.shape[0]

egc_index_Y_to_X_avg = egc_index_Y_to_X.mean(axis=0)
egc_index_Y_to_X_err = egc_index_Y_to_X.std(axis=0) / egc_index_Y_to_X.shape[0]
```

```python
plt.plot(epsilons, egc_index_X_to_Y_avg, markers[0], color=colors[0], label="$X\\rightarrow Y$")
plt.plot(epsilons, egc_index_Y_to_X_avg, markers[1], color=colors[1], label="$Y\\rightarrow X$")
plt.fill_between(epsilons, egc_index_X_to_Y_avg - egc_index_X_to_Y_err,
                           egc_index_X_to_Y_avg + egc_index_X_to_Y_err, color=colors[0], alpha=0.4)
plt.fill_between(epsilons, egc_index_Y_to_X_avg - egc_index_Y_to_X_err,
                           egc_index_Y_to_X_avg + egc_index_Y_to_X_err, color=colors[1], alpha=0.4)
plt.ylabel("Imbalance Gain [%]")
#plt.set_xticks([])
#plt.legend()
```

```python
for seed in range(20):
    plt.plot(epsilons, egc_index_X_to_Y[seed], '-', color=colors[0], label="$X\\rightarrow Y$", alpha=0.2)
    plt.plot(epsilons, egc_index_Y_to_X[seed], '-', color=colors[1], label="$Y\\rightarrow X$", alpha=0.2)
plt.ylabel("Imbalance Gain [%]")
#plt.set_xticks([])
#plt.legend()
```
