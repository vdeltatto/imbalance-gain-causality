```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
import pickle
import scienceplots

plt.style.use(['science','no-latex', 'notebook'])
colors = ["#00BBF9", "#CC2836", "#F3A712","#3eb690","#A05496","#114b5f"]
markers = ['o-','s-','^-']
#mpl.rcParams.update(mpl.rcParamsDefault)
```

```python
def compute_IG(imbalances):
    return (imbalances[:,0] - np.min(imbalances, axis=-1)) / imbalances[:,0]
```

### Lorenz 96, all coordinates

```python
epsilons_zx = np.arange(0.,1.5+0.1,0.1)
print(f"epsilons_zx = {epsilons_zx}")
alphas = np.linspace(0.,1.5,200)

# X and Y
info_imbalances_X_to_Y = np.zeros((len(epsilons_zx),len(alphas)))
info_imbalances_Y_to_X = np.zeros((len(epsilons_zx),len(alphas)))
for i_eps, eps in enumerate(epsilons_zx):
    info_imbalances_X_to_Y[i_eps], info_imbalances_Y_to_X[i_eps] = pickle.load(open(f"./ig_allcoords/pickles/lorenz96/XY_ieps{i_eps}.p","rb"))
IGs_X_to_Y = compute_IG(info_imbalances_X_to_Y)
IGs_Y_to_X = compute_IG(info_imbalances_Y_to_X)

# Z and Y
info_imbalances_Z_to_Y = np.zeros((len(epsilons_zx),len(alphas)))
info_imbalances_Y_to_Z = np.zeros((len(epsilons_zx),len(alphas)))
for i_eps, eps in enumerate(epsilons_zx):
    info_imbalances_Z_to_Y[i_eps], info_imbalances_Y_to_Z[i_eps] = pickle.load(open(f"./ig_allcoords/pickles/lorenz96/ZY_ieps{i_eps}.p","rb"))
IGs_Z_to_Y = compute_IG(info_imbalances_Z_to_Y)
IGs_Y_to_Z = compute_IG(info_imbalances_Y_to_Z)
```

```python
plt.figure(figsize=(7,5))
plt.plot(epsilons_zx, 100*IGs_X_to_Y, 'o-', label="$X\\rightarrow Y$")
plt.plot(epsilons_zx, 100*IGs_Y_to_X, 's-', label="$Y\\rightarrow X$")
plt.plot(epsilons_zx, 100*IGs_Z_to_Y, 'd-', label="$Z\\rightarrow Y$")
plt.plot(epsilons_zx, 100*IGs_Y_to_Z, '^-', label="$Y\\rightarrow Z$")
plt.xlabel("$\\epsilon_{zx}$")
plt.ylabel("Imbalance Gain [%]")
plt.legend()
plt.savefig("./figures/allcoords.pdf", dpi=300)
plt.show()
```
