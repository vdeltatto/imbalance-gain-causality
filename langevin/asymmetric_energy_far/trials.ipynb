```python
import numpy as np
import knncmi
from pyitlib import discrete_random_variable as drv
import pandas as pd
```

### Check that TE measures give 0 at tau=0

```python
# continuous TE
X0 = np.random.normal(size=(1000,1))
Y0 = np.random.normal(size=(1000,1))
Xtau = +X0
Ytau = +Y0

variable_names = ['x_0', 'y_0', 'x_tau', 'y_tau']
dataset = pd.DataFrame(np.column_stack((X0,Y0,Xtau,Ytau)), columns=variable_names)

TEs_X_to_Y = (
    knncmi.cmi(['x_0'], ['y_tau'], ['y_0'], k=7, data=dataset, minzero=False)
)
print(TEs_X_to_Y==0)

# discrete TE
X0 = np.array(np.random.normal(size=(1000)), dtype=int)
Y0 = np.array(np.random.normal(size=(1000)), dtype=int)
Xtau = +X0
Ytau = +Y0

TEs_X_to_Y_discrete = (
    drv.information_mutual_conditional(X=X0,Y=Ytau,Z=Y0, base=np.exp(1))
)
print(TEs_X_to_Y_discrete==0)
```
