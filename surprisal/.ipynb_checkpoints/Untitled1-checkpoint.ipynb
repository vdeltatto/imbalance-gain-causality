```python
import numpy as np
import matplotlib.pyplot as plt
import pickle
from sklearn.neighbors import NearestNeighbors
import pandas as pd
```

```python
def autocorr2(x): # autocorrelation function
    r2=np.fft.ifft(np.abs(np.fft.fft(x))**2).real
    c=(r2/x.shape-np.mean(x)**2)/np.std(x)**2
    return c[:len(x)//2]
```

```python
plt.plot(autocorr2(brain_data['PostTemp'])[:10],'o-')
```

```python
times = np.linspace(0,1210,1210,dtype=int)
nsteps = len(times)

E = 4
trajectory_x = brain_data['PostTemp'][times]#.reshape(-1, 1)

indices_nn_x = np.zeros(nsteps)

nbrs_x = NearestNeighbors(n_neighbors=nsteps).fit(trajectory_x)
    
_, indices_nn_x = nbrs_x.kneighbors(trajectory_x)
    
plt.rcParams['font.size'] = 16
fig, ax = plt.subplots(figsize=(5,5))
ax.plot(indices_nn_x[:,1],'o')
ax.set(xlabel="time step", ylabel="nn index", title="X trajectory")
plt.show()
```
