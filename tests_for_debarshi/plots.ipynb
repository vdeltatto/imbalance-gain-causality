```python
import numpy as np
import matplotlib.pyplot as plt
import pickle
```

## PLOTS OF PLAIN II(tor1,tor2 -> CNs)

### k = 1

```python
k = 1

seeds = np.arange(1,15)
coords = [0,1,2,3,4,"all"]
taus = np.linspace(0,100,21,dtype=int)

imbs = {}

for coord in coords:
    imbs[coord] = np.zeros((len(seeds),len(taus)))
    for iseed, seed in enumerate(seeds):
        with open(f"/scratch/vdeltatt/imbalance-gain-causality/tests_for_debarshi/pickles_ii/seed{seed}_coord{coord}_k{k}.p","rb") as f:
            taus, imbs_temp = pickle.load(f)
            imbs[coord][iseed] = +imbs_temp[:,0]
```

```python
dict_cns = {0: "cn_NT", 1: "cn_CT", 2: "cn_O1", 3: "cn_O2", 4: "cn_NE1", "all": "all cns"}
```

```python
for coord in coords:
    plt.figure(figsize=(7,3))
    avg = imbs[coord].mean(axis=0)
    err = imbs[coord].std(axis=0, ddof=1) / np.sqrt(imbs[coord].shape[0])
    plt.plot(taus, avg, 'o-', label=dict_cns[coord])
    plt.fill_between(taus, avg - err, avg + err, alpha = 0.4)
    plt.hlines(1.0, taus[0], taus[-1], linestyle="--", color="black")

    plt.legend()
    plt.ylabel("Information Imbalance")
    plt.xlabel("$\\tau$")
    plt.show()
```

## PLOTS OF IG

```python
def compute_IGs(imbalances, correction=True):    
    n_indep_realizations = imbalances.shape[0]
    n_taus = imbalances.shape[1]
    
    IGs_vs_alpha = 100*(imbalances[:,:,0,np.newaxis]-imbalances) / imbalances[:,:,0,np.newaxis]
    alphas_max = np.argmax(IGs_vs_alpha.mean(axis=0), axis=-1)

    IGs = IGs_vs_alpha.mean(axis=0)[np.arange(n_taus),alphas_max]
    errors = IGs_vs_alpha.std(axis=0)[np.arange(n_taus),alphas_max] / np.sqrt(n_indep_realizations)

    if correction:
        IGs[imbalances[:,:,0].mean(axis=0) > 1] = 0
        errors[imbalances[:,:,0].mean(axis=0) > 1] = 0
        
    return IGs, errors
```

### k = 5

```python
k = 5

seeds = np.arange(1,15)
coords = [0,1,2,3,4,"all"]
taus = np.linspace(0,100,21,dtype=int)
alphas = np.linspace(0,20,500)

imbs_XtoY = {}
imbs_YtoX = {}

for coord in coords:
    imbs_XtoY[coord] = np.zeros((len(seeds),len(taus),len(alphas)))
    imbs_YtoX[coord] = np.zeros((len(seeds),len(taus),len(alphas)))
    for iseed, seed in enumerate(seeds):
        with open(f"/scratch/vdeltatt/imbalance-gain-causality/tests_for_debarshi/pickles_ig/seed{seed}_coord{coord}_k{k}.p","rb") as f:
            taus, alphas, imbs_XtoY[coord][iseed], imbs_YtoX[coord][iseed] = pickle.load(f)

IGs_XtoY = {}
IGs_YtoX = {}
errs_XtoY = {}
errs_YtoX = {}
for coord in coords:
    IGs_XtoY[coord], errs_XtoY[coord] = compute_IGs(imbs_XtoY[coord], correction=False)
    IGs_YtoX[coord], errs_YtoX[coord] = compute_IGs(imbs_YtoX[coord], correction=False)
```

```python
dict_cns = {0: "cn_NT", 1: "cn_CT", 2: "cn_O1", 3: "cn_O2", 4: "cn_NE1", "all": "all cns"}
```

```python
for coord in coords:
    plt.figure(figsize=(7,3))
    plt.plot(taus, IGs_XtoY[coord], 'o-', label=f"tor -> {dict_cns[coord]}")
    plt.fill_between(taus, IGs_XtoY[coord] - errs_XtoY[coord], IGs_XtoY[coord] + errs_XtoY[coord], alpha = 0.4)
    plt.plot(taus, IGs_YtoX[coord], 'o-', label=f"{dict_cns[coord]} -> tor")
    plt.fill_between(taus, IGs_YtoX[coord] - errs_YtoX[coord], IGs_YtoX[coord] + errs_YtoX[coord], alpha = 0.4)
    #plt.hlines(1.0, taus[0], taus[-1], linestyle="--", color="black")

    plt.legend()
    plt.ylabel("Information Imbalance")
    plt.xlabel("$\\tau$")
    plt.show()
```
