```python
import numpy as np
import matplotlib.pyplot as plt
import pickle
import scienceplots

plt.style.use(['science','notebook'])
colors = ["#00BBF9", "#CC2836", "#F3A712","#3eb690","#A05496","#114b5f"]
markers = ['o-','s-','^-']
```

### Lorenz

```python
Es = np.arange(1,15)
info_imbalances = np.zeros((20,len(Es),2))

for seed in range(20):
    info_imbalances[seed] = pickle.load(open(f"./pickles/lorenz/seed{seed}.p","rb"))
print(info_imbalances.shape)
```

```python
fig, ax = plt.subplots(1,2,figsize=(12,5))
ax[0].plot(Es, info_imbalances[:,:,0].mean(axis=0), markers[0], color=colors[0], label="$E\\rightarrow E+1$")
ax[0].plot(Es, info_imbalances[:,:,1].mean(axis=0), markers[1], color=colors[1], label="$E+1\\rightarrow E$")
#plt.fill_between(epsilons, IGs_X_to_Y-errors_X_to_Y, IGs_X_to_Y+errors_X_to_Y, color=colors[0], alpha=0.4)
#plt.fill_between(epsilons, IGs_Y_to_X-errors_Y_to_X, IGs_Y_to_X+errors_Y_to_X, color=colors[1], alpha=0.4)
ax[0].set(ylabel="Information Imbalance", xlabel="Embedding length E", yscale="log", title="Absolute info imbalance")
ax[0].legend()

diff_forward = (info_imbalances[:,:-1,0].mean(axis=0) - info_imbalances[:,1:,0].mean(axis=0)) / info_imbalances[:,:-1,0].mean(axis=0)
diff_backward = (info_imbalances[:,:-1,1].mean(axis=0) - info_imbalances[:,1:,1].mean(axis=0)) / info_imbalances[:,:-1,1].mean(axis=0)
ax[1].plot(Es[:-1], diff_forward, markers[0], color=colors[0], label="$E\\rightarrow E+1$")
ax[1].plot(Es[:-1], diff_backward, markers[1], color=colors[1], label="$E+1\\rightarrow E$")
#plt.fill_between(epsilons, IGs_X_to_Y-errors_X_to_Y, IGs_X_to_Y+errors_X_to_Y, color=colors[0], alpha=0.4)
#plt.fill_between(epsilons, IGs_Y_to_X-errors_Y_to_X, IGs_Y_to_X+errors_Y_to_X, color=colors[1], alpha=0.4)
ax[1].set(ylabel="Information Imbalance", xlabel="Embedding length E", title="Relative diff in info imbalance", yscale="log")
ax[1].legend()

plt.show()
```
